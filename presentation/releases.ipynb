## RELEASES

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
```

```python
img_extent = [65.0, 95.0, 5.0, 35.0]
releases = {'Delhi': [77.2, 28.6], 'Khammam': [80.1, 17.2], 'Karachi': [67.1, 24.9]}
```

```python
fig, ax = plt.subplots(1, 1, figsize=(6, 6),
                       subplot_kw={'projection': ccrs.PlateCarree()}
                      )

for key in ['Delhi', 'Karachi']:
    ax.scatter(releases[key][0], releases[key][1], c='white', edgecolor='black', marker='*', s=250, zorder=5, transform=ccrs.PlateCarree(), label=key + f' (Lon: {releases[key][0]}°, Lat: {releases[key][1]}°)')
    ax.annotate(key, (releases[key][0], releases[key][1]), xytext=(10, -10), textcoords='offset points', color='black', weight='bold', fontsize=10)
    
ax.set_extent(img_extent)
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS, lw=0.5, alpha=0.8)
ax.add_feature(cfeature.LAND)
ax.add_feature(cfeature.OCEAN)
ax.legend()

gl = ax.gridlines(draw_labels=True, color='gray', alpha=0.5, linestyle='--')
gl.top_labels = False
gl.right_labels = False

fig.savefig('Releases.pdf')
```
