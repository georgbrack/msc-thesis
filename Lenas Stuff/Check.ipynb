## Looking at User-defined Initial Conditions

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.ticker as ticker
import cartopy.crs as ccrs
import cartopy.feature as cfeature
```

```python
part_ic_j = xr.open_dataset("part_ic_June.nc")
```

```python
part_ic_j
```

```python
fig, ax = plt.subplots(1, 1, layout='compressed', subplot_kw={'projection': ccrs.PlateCarree()})

scatter = ax.scatter(part_ic_j.longitude, part_ic_j.latitude)

gl = ax.gridlines(draw_labels=True, color='gray', alpha=0.5, linestyle='--')
gl.top_labels = False
gl.right_labels = False

ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS, lw=0.2)
```
